# Practical work

````python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import spacy
import re
import pandas as pd
import matplotlib.pyplot as plt
````

## Load data

````python
bbc_data = pd.read_csv("bbc_news.csv")
````

````python
bbc_data.head()
````

````python
bbc_data.info()
````

````python
titles = pd.DataFrame(bbc_data["title"])
````

````python
titles.head()
````

````python
# lowercase
titles['lowercase'] = titles['title'].str.lower()
````

````python
# stop words removal
en_stopwords = stopwords.words("english")
titles['no_stopwords'] = titles.apply(lambda s: ' '.join([word for word in s["lowercase"].split() if word not in en_stopwords]), axis=1 )
````

````python
# punctuation removal
titles['no_stopwords_no_punct'] = titles['no_stopwords'].apply(lambda s: re.sub(r"[^\w\s]", "", s))
````

````python
# tokenize
titles["tokens_raw"] = titles.apply(lambda s: word_tokenize(s["title"]), axis=1)
titles["tokens_clean"] = titles.apply(lambda s: word_tokenize(s["no_stopwords_no_punct"]), axis=1)

````

````python
# lemmatizing
lemmatizer = WordNetLemmatizer()
titles['tokens_clean_lemmatized'] = titles['tokens_clean'].apply(lambda tokens: [lemmatizer.lemmatize(token) for token in tokens])
````

````python
titles.head(3)
````

````python
# create lists for just our tokens
tokens_raw_list = sum(titles['tokens_raw'], [])
tokens_clean_list = sum(titles['tokens_clean_lemmatized'], [])
````

### **POS Tagging**

````python
nlp = spacy.load("en_core_web_sm")
````

````python
spacy_doc = nlp(' '.join(tokens_raw_list))
````

````python
pos_df = pd.DataFrame(
    columns=['token', 'pos_tag']
)
````

````python
for token in spacy_doc:
    pos_df = pd.concat(
        [pos_df, pd.DataFrame.from_records([{"token": token.text, "pos_tag": token.pos_}])], ignore_index=True)
````

````python
pos_df_counts = pos_df.groupby(["token", "pos_tag"]).size().reset_index(name="counts").sort_values(by="counts", ascending=False)
````

````python
pos_df_counts
````

# **NER**

````python
ner_df = pd.DataFrame(columns=['token', 'ner_tag'])

for token in spacy_doc.ents:
    ner_df = pd.concat(
        [ner_df, pd.DataFrame.from_records([{"token": token.text, "ner_tag": token.label_}])], ignore_index=True
    )
````

````python
ner_df_counts = ner_df.groupby(["token", "ner_tag"]).size().reset_index(name="counts").sort_values(by="counts", ascending=False)
````
