# Stemming

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
connect_tokens = ["connecting", "connected", "connectivity", "connect", "connects"]
```

```python
for t in connect_tokens:
    print(t, ": ", ps.stem(t))
```

```python
learn_tokens = ["learned", "learning", "learn", "learns", "learner", "learners"]
```

```python
for t in learn_tokens:
    print(t, ": ", ps.stem(t))
```

```python
likes_tokens = ["likes", "better", "worse"]
```

```python
for t in likes_tokens:
    print(t, ": ", ps.stem(t))
```

# All togather

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
import re
import pandas as pd
```

```python
data = pd.read_csv("tripadvisor_hotel_reviews.csv")
```

```python
data.info()
```

```python
data.head()
```

```python
data["Review"][0]
```

```python
data["review_lowercase"] = data["Review"].str.lower()
```

```python
data.head()
```

```python
en_stopwords = stopwords.words("english")
```

```python
en_stopwords.remove("not")
```

```python
data["review_no_stopwords"] = data['review_lowercase'].apply(lambda sen: " ".join((word for word in sen.split() if word not in en_stopwords)))
```

```python
data.head()
```

```python
data["review_no_stopwords"][0]
```

```python
data["review_no_stopwords_no_punct"] = data.apply(lambda x: re.sub(r'[*]', "star", x["review_no_stopwords"]), axis=1 )
```

```python
data.head()
```

### Removing punctuations #1

```python
data["review_no_stopwords_no_punct"] = data.apply(lambda x: re.sub(r"[^\w\s]", "", x["review_no_stopwords_no_punct"]), axis=1)
```

```python
data.head()
```

```python
data.head()
```

### Removing punctuations #2

```python
data["review_no_stopwords_no_punct"] = data["review_no_stopwords_no_punct"].apply(lambda x: re.sub(r"[^\s\w]", "", x))
```

```python
data.head()
```

```python
data["tokenized"] = data.apply(lambda s: word_tokenize(s['review_no_stopwords_no_punct']) , axis=1)
```

```python
data.head()
```

```python
data['tokenized'][0]
```

```python
ps = PorterStemmer()
```

```python
data["stemmed"] = data["tokenized"].apply(lambda tokens: [ps.stem(token) for token in tokens])
```

```python
data.head(20)
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
data["lemmatized"] = data["tokenized"].apply(lambda tokens: [lemmatizer.lemmatize(token) for token in tokens])
```

```python
data.head()
```

```python
tokens_clean = sum(data['lemmatized'], [])
```

```python
tokens_clean
```

```python
unigrams = (pd.Series(nltk.ngrams(tokens_clean, 1)).value_counts())
print(unigrams)
```

```python
bigrams = (pd.Series(nltk.ngrams(tokens_clean, 2)).value_counts())
print(bigrams)
```
